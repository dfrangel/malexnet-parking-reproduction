```python
import sys
import os
module_path = os.path.abspath(os.path.join('..', 'src'))
if module_path not in sys.path:
    sys.path.append(module_path)
from dataloader import Dataset
from PIL import Image
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader

dataset = Dataset(
    patches_path="../data/CNRPark-EXT/PATCHES",
    labels_file="../data/CNRPark-EXT/LABELS/train.txt",
)

batch_size = 5

train_dataloader_img = DataLoader(dataset.img_paths, batch_size, shuffle=None)
train_dataloader_label = DataLoader(dataset.img_labels, batch_size, shuffle=None)
```

```python
# Display image and label.
train_features = next(iter(train_dataloader_img))
train_labels = next(iter(train_dataloader_label))
print(f"Feature batch shape: {len(train_features)}")
print(f"Labels batch shape: {len(train_labels)}")

# Print all the images and labels (TRY WITH SMALL BATCHES)
for i in range(0, len(train_features)):
    img = Image.open(train_features[i])
    label = train_labels[i]
    plt.imshow(img)
    plt.show()
    print(f"Label: {label}")
```
