```python
import sys
import os
module_path = os.path.abspath(os.path.join('..', 'src'))
if module_path not in sys.path:
    sys.path.append(module_path)
from dataloader import Dataset
from PIL import Image
import matplotlib.pyplot as plt
from torch.utils.data import DataLoader

dataset = Dataset(
    patches_path="../data/CNRPark-EXT/PATCHES",
    labels_file="../data/CNRPark-EXT/LABELS/train.txt",
)

batch_size = 64

train_dataloader_img = DataLoader(dataset.img_paths, batch_size=50, shuffle=None)
train_dataloader_label = DataLoader(dataset.img_labels, batch_size=50, shuffle=None)
```
